## Student Details
Run this cell first. The rest of the notebook uses these values.

```python
roll_no = input("Enter your 8-digit roll number: ").strip()

if not roll_no.isdigit() or len(roll_no) != 8:
    raise ValueError("Enter exactly 8 digits.")

name = input("Enter your name: ")
branch = input("Enter your branch: ")
age = int(input("Enter your age: "))
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

digits = [int(x) for x in roll_no]

print("Roll number:", roll_no)
print("Digits:", digits)
```

## Q1. Lists

Take the roll number, extract its individual digits, and multiply each digit by 10 to form list `L`.

```python
L = [x * 10 for x in digits]

print("L =", L)

L.append(55)
print("After append:", L)

L.insert(2, 75)
print("After insert:", L)

L.remove(55)
print("After remove:", L)

L.pop()
print("After pop:", L)

L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

print("First three:", L[:3])
print("Last three:", L[-3:])

avg = sum(L) / len(L)
new_L = [x for x in L if x > avg]

print("Average:", avg)
print("Above average:", new_L)
```

## Q2. Tuples

Create `scores` using the first 8 values from the original digit-based list.

```python
base_L = [x * 10 for x in digits]
scores = tuple(base_L[:8])

print("scores =", scores)

highest = max(scores)
lowest = min(scores)

print("Highest score:", highest)
print("Index of highest:", scores.index(highest))
print("Lowest score:", lowest)
print("Times lowest occurs:", scores.count(lowest))

print("Reversed tuple as list:", list(reversed(scores)))

score = int(input("Enter a score: "))

if score in scores:
    print("First occurrence index:", scores.index(score))
else:
    print("Score is not present")

try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)

first, second, *remaining = scores
print("First:", first)
print("Second:", second)
print("Remaining:", remaining)
```

## Q3. Random Numbers

The random seed is set to your roll number so the output is reproducible for you.

```python
import random
from collections import Counter

random.seed(int(roll_no))

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

odd = [x for x in numbers if x % 2 != 0]
even = [x for x in numbers if x % 2 == 0]

print("Odd count:", len(odd))
print("Odd numbers:", odd)

print("Even count:", len(even))
print("Even numbers:", even)

def prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

primes = [x for x in numbers if prime(x)]

print("Prime count:", len(primes))
print("Prime numbers:", primes)

count = Counter(numbers)
most_common = count.most_common(1)[0]

print("Most frequent number:", most_common[0])
print("Times it occurs:", most_common[1])
```

## Q4. Sets

Use the same 8 original digits (before multiplying by 10) to create sets A and B.

```python
A = {x * 7 for x in digits}
B = {x * 9 for x in digits}

print("A =", A)
print("B =", B)

print("Union:", A.union(B))
print("Intersection:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Symmetric difference:", A.symmetric_difference(B))

print("A subset of B:", A.issubset(B))
print("B superset of A:", B.issuperset(A))

x = int(input("Enter a value: "))
A.discard(x)

print("A after discard:", A)
```

## Q5. Dictionaries

Create the dictionary with the student's details and perform the required operations.

```python
my_dict = {
    "name": name,
    "roll_no": roll_no,
    "branch": branch,
    "age": age,
    "city": city
}

print("Original dictionary:", my_dict)

original_dict = my_dict.copy()

my_dict["location"] = my_dict.pop("city")
print("After renaming city:", my_dict)

my_dict["cgpa"] = cgpa
print("After adding cgpa:", my_dict)

my_dict["age"] += 1
print("After increasing age:", my_dict)

dict1 = my_dict.copy()
removed = dict1.pop("branch")
print("Using pop():", dict1)
print("Removed value:", removed)

dict2 = my_dict.copy()
del dict2["branch"]
print("Using del:", dict2)

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email is not present")

friend_dict = {
    "name": "Rahul",
    "roll_no": "99999999",
    "branch": branch,
    "age": age,
    "city": "Delhi"
}

merged = {**original_dict, **friend_dict}

print("Friend dictionary:", friend_dict)
print("Merged dictionary:", merged)

string_dict = {
    key: value for key, value in original_dict.items()
    if isinstance(value, str)
}

print("String values:", string_dict)
```
