## Assignment 5.3

**Question:** WAP to add all prime numbers from 1 to n, where n is given by the user.

```python
n = int(input("Enter n: "))

total = 0

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        total += num

print("Sum of prime numbers =", total)
```

## Assignment 6.1

**Question:** WAP using a function to add all odd numbers from 1 to n, where n is given by the user.

```python
def sum_odd_numbers(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total


n = int(input("Enter n: "))
print("Sum of odd numbers =", sum_odd_numbers(n))
```

## Assignment 6.2

**Question:** WAP using a function to add all prime numbers from 1 to n, where n is given by the user.

```python
def sum_prime_numbers(n):
    total = 0

    for num in range(2, n + 1):
        is_prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            total += num

    return total


n = int(input("Enter n: "))
print("Sum of prime numbers =", sum_prime_numbers(n))
```
