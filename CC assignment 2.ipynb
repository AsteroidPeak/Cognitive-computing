```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df2 = pd.DataFrame(data)

df2.to_csv("employees.csv", index=False)

print(df2)
```

```python
print(df2.shape)
```

```python
print(df2.info)
```

```python
print(df2.describe)
```

```python
print(df2.head(5),df2.tail(3))
```

```python
print("Average salary:", df2["Salary"].mean())

print("Total bonus:", df2["Bonus"].sum())

print("Youngest age:", df2["Age"].min())

print("Highest rating:", df2["Rating"].max())
```

```python
df2 = df2.sort_values(by="Salary", ascending=False)

print("\nDataFrame sorted by Salary (descending):")
print(df2)
```

```python
df2["Performance"] = df2["Rating"].apply(
    lambda x: "Excellent" if x >= 4.5
    else "Good" if x >= 4.0
    else "Average"
)

print("\nDataFrame with Performance column:")
print(df2)
```

```python
print("\nMissing values in each column:")
print(df2.isnull().sum())
```

```python
df2.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\nDataFrame after renaming Employee_ID to ID:")
print(df2)
```

```python
experienced = df2[df2["Years_of_Experience"] > 5]

print("\nEmployees with more than 5 years of experience:")
print(experienced)

it_employees = df2[df2["Department"] == "IT"]

print("\nEmployees belonging to IT department:")
print(it_employees)
```

```python
df2["Tax"] = df2["Salary"] * 0.10

print("\nDataFrame with Tax column:")
print(df2)
```

```python
df2.to_csv("employees_modified.csv", index=False)

print("\nModified DataFrame saved as 'employees_modified.csv'")
```
