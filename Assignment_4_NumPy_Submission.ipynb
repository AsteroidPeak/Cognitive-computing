```
import numpy as np
```

## Q1

```
arr=np.array([1,2,3,4,5])
print(arr+2)
print(arr*3)
print(arr/2)
```

## Q2

```
arr=np.array([1,2,3,6,4,5])
print(arr[::-1])

x=np.array([1,2,3,4,5,1,2,1,1,1])
v=np.bincount(x).argmax()
print(v,np.where(x==v)[0])

y=np.array([1,1,1,2,3,4,2,4,3,3])
v=np.bincount(y).argmax()
print(v,np.where(y==v)[0])
```

## Q3

## Q4

```
Ansh=np.linspace(10,100,25)
print(Ansh)
print("Dimensions:",Ansh.ndim)
print("Shape:",Ansh.shape)
print("Elements:",Ansh.size)
print("Datatype:",Ansh.dtype)
print("Bytes:",Ansh.nbytes)
t=Ansh.reshape(25,1)
print(t)
print(t.T)
```

## Q5

```
ucs420_Ansh=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print("Mean:",np.mean(ucs420_Ansh))
print("Median:",np.median(ucs420_Ansh))
print("Max:",np.max(ucs420_Ansh))
print("Min:",np.min(ucs420_Ansh))
print("Unique:",np.unique(ucs420_Ansh))
reshaped_ucs420_Ansh=ucs420_Ansh.reshape(4,3)
print(reshaped_ucs420_Ansh)
resized_ucs420_Ansh=np.resize(ucs420_Ansh,(2,3))
print(resized_ucs420_Ansh)
```
