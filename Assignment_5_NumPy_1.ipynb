```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
print(arr)
print(arr + 2)
print(arr * 3)
print(arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

for values in [x, y]:
    unique, counts = np.unique(values, return_counts=True)
    frequent_value = unique[np.argmax(counts)]
    print(frequent_value)
    print(np.where(values == frequent_value)[0])
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print(arr[0, 1])
print(arr[2, 0])
```

```python
ansh_sharma = np.linspace(10, 100, 25)
print(ansh_sharma)
print(ansh_sharma.ndim)
print(ansh_sharma.shape)
print(ansh_sharma.size)
print(ansh_sharma.dtype)
print(ansh_sharma.nbytes)
print(ansh_sharma.reshape(25, 1))
print(ansh_sharma.T)
```

```python
ucs420_ansh_sharma = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print(ucs420_ansh_sharma)
print(np.mean(ucs420_ansh_sharma))
print(np.median(ucs420_ansh_sharma))
print(np.max(ucs420_ansh_sharma))
print(np.min(ucs420_ansh_sharma))
print(np.unique(ucs420_ansh_sharma))
reshaped_ucs420_ansh_sharma = ucs420_ansh_sharma.reshape(4, 3)
print(reshaped_ucs420_ansh_sharma)
resized_ucs420_ansh_sharma = np.resize(ucs420_ansh_sharma, (2, 3))
print(resized_ucs420_ansh_sharma)
```
